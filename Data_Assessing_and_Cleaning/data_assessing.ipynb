## Data Accessing

Types of unclean data:

1. Dirty Data (Data with Quality Issues): Dirty data, also known as low quality data. Low quality data has content issues.
    - Duplicated data
    - Missing data
    - Corrupt data
    - Inaccurate data

2. Messy Data (Data with tidiness issues): Messy data, also known as untidy data. Untidy data has structural issues.

    Tidy data has the following properties:
    - Each variable forms a column
    - Each observation forms a row
    - Each observational unit forms a table
