# NumPy

A Basic Knowledge of NumPy.

NumPy (Numerical Python) is an open-source library that provides support for large, multi-dimensional arrays and matrices, as well as a wide range of mathematical functions to operate on these arrays.

At its core, NumPy provides the ndarray (n-dimensional array) data structure, which is highly optimized for efficient mathematical operations. This data structure allows you to store and manipulate large sets of data in a homogeneous and memory-efficient way. 

NumPy is a low-level library written in C for high-level mathematical functions. 
NumPy cleverly overcomes the problem of running slower algorithms on Python using multidimensional arrays and functions that operate on arrays. Any algorithm can then be expressed as a function on these arrays, allowing the algorithms to be run quickly.

Applications of NumPy

- A powerful N-dimensional array object.
- Sophisticated functions.
- Tools for integrating C/C++ and Fortran code.
- Useful linear algebra, Fourier transforms, and random number capabilities.

N-d Array Properties

- a powerful N-dimensional array object
- sophisticated (broadcasting) functions
- tools for integrating C/C++ and Fortran code
- useful linear algebra, Fourier transform, and random number capabilities

Properties of NumPy

- Homogenous (Only one type of data can be stored unlike Python Lists)
- Fixed item size (You can't directly add new elements to an array without creating a new one)

How n-D Array's are better than Lists for large numerical datasets

1. Faster
2. Convenient
3. Less Memory

```python
import sys
list_a = list(range(100))
np_arr = np.arange(100)
```

```python
print(sys.getsizeof(87)*len(list_a)) 
#getsizeof tells the memory consumed by a particular item, 
# here we multiplied that with total number of such items, i.e., 100.
```

# Array Operations

```python
a1 = np.arange(12).reshape(3,4)
a2 = np.arange(12,24).reshape(3,4)
```

### Scalar Operation


```python
# Arithemetic operation every element with a scaler quantity (basically a number).

print(a1 + 4,"\n")
print(a1 - 1,"\n")
print(a1 * 3,"\n")
print(a1 / 4,"\n")
print(a1 // 5,"\n")
print(a1 % 9,"\n")
print(a1 ** 2,"\n")
```

```python
# Relational Operations - Comparision of every element with a scalar.

print(a1 > 1, '\n')
print(a1 >= 3, '\n')
print(a2 < 15, '\n')
print(a2 <= 20, '\n')
print(a1 == 5, '\n')
print(a1 != 2, '\n')
```

### Vector Operations (Shape must be same for both)

```python
# Arithemetic operation - Element wise operations.
print(a1 +  a2,"\n")
print(a1 -  a2,"\n")
print(a1 *  a2,"\n")
print(a1 /  a2,"\n")
print(a1 // a2,"\n")
print(a1 %  a2,"\n")
print(a2 ** a1,"\n")
```

```python
# Relational Operations - Elementwise comparision of every element.

print(a1 >  a2, '\n')
print(a1 >= a2, '\n')
print(a2 <  a2, '\n')
print(a2 <= a2, '\n')
print(a1 == a2, '\n')
print(a1 != a2, '\n')
```

### Mathematical Operations - Min / Max / Sum / Prod

```python
# For the entire tensor
print(np.max(a1))
print(np.min(a1))
print(np.sum(a2))
print(np.prod(a2))
```

```python
# For each row/column, not entire tensor.
# Use axis, 0 means columns, 1 means rows

print(np.max(a1 ,axis = 0),"\n")
print(np.min(a1 ,axis = 0),"\n")
print(np.sum(a1 ,axis = 0),"\n")
print(np.prod(a1,axis = 0),"\n")
```

### Statistical Operation - Mean / Median / Std / Var

```python
# For the entire tensor

print(np.mean(a1))
print(np.median(a1))
print(np.std(a2))
print(np.var(a2))
```

```python
# For each row/column, not entire tensor.
# Use axis, 0 means columns, 1 means rows

print(np.mean(a1 ,axis = 0),"\n")
print(np.median(a1 ,axis = 0),"\n")
print(np.std(a1 ,axis = 0),"\n")
print(np.var(a1,axis = 0),"\n")
```

### Trignometric / Logs / Exponents / Dot Product

```python
print(np.sin(a1),"\n")

print(np.log(a2),"\n")

print(np.exp(a1),"\n")

# print(np.dot(a1,a2)) 
# # This will fail cause it doesn't obeys the dot product rule.
```

### Round / Floor / Ceil

```python
a3 = np.random.random((2,3))*100
print(a3,"\n")

print(np.round(a3,decimals=2),"\n") #rounds upto a desired num of decimal place.

print(np.floor(a3),"\n") #Removes the decimal part without changing whole.

print(np.ceil(a3),"\n") #Next Heightest integer
```

# Indexing, Slicing and Iterating

### Indexing
Fetching out a single element.

```python
a1 = np.arange(10)
a2 = np.arange(12).reshape(3,4)
a3 = np.arange(27).reshape(3,3,3)
```

```python
# 1-D Vector
print(a1)      #prints array
print(a1[0])   #prints first index, left to right
print(a1[-1])  #prints last  index, right to left
```

```python
# 2-D Matrix
print(a2)
#We have to tell the row and coloumn to get the desired num.
print(a2[2,2]) 
```

```python
# 3-D Tensor
print(a3)

#We have to provide the right order.
#Solve form higher to lower.

print(a3[1,0,1]) 
print(a3[0,1,1])
print(a3[0,0,0]) 
#Same applies to higher dimensional arrays.
```

### Slicing
Fetching multiple elements.

```python
# 1-D Vectors
print(a1)

print(a1[2:8]) #Slicing a given range.
print(a1[5:])  #Slicing from one side.
print(a1[2:8:2]) #Step size - skip every second element
print(a1[8:1:-1]) #Step size - reverse slicing
```

```python
# 2-D Matrix

print(a2[:,:],'\n')

print(a2[:,2],'\n')

print(a2[2:3,:],'\n')

print(a2[:,1:2],'\n')

print(a2[1:3,1:3],'\n')

print(a2[::2,::3],'\n')

print(a2[::2,1::2])
```

```python
# 3-D Matrix
print(a3,'_______________________',sep='\n')
print(a3[1,:,:],'_______________________',sep='\n') #slice the middle array
print(a3[::2],'_______________________',sep='\n') #slice the except middle array
print(a3[0,1],'_______________________',sep='\n') #second row of 1st array
print(a3[1,1,1],'_______________________',sep='\n') #middle element of second row of 2nd array
print(a3[2,1:,1:],'_______________________',sep='\n') #22 23 25 26
```

### Iterating

```python
print(a2)
for i in np.nditer(a2):
    print(i)
```

# Reshaping

```python
# Reshape - Shape your array

arr1 = np.arange(12).reshape(3,4)  # Number of elements must be a multiple of shape values.
arr2 = np.arange(12).reshape(2,-1) # -1 offset will figure out the other value.
print(arr1)
print(arr2)
```

```python
# Transpose - Matrix Transpose, switches rows and columns

print(a2,'\n')
print(a2.T,'\n')
print(np.transpose(a2),'\n')
```

```python
# Ravel - Brings higher dimension arrays to 1-d array

print(a3,'\n')

print(np.ravel(a3))
```
