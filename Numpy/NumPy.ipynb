# NumPy

A Basic Knowledge of NumPy.

NumPy (Numerical Python) is an open-source library that provides support for large, multi-dimensional arrays and matrices, as well as a wide range of mathematical functions to operate on these arrays.

At its core, NumPy provides the ndarray (n-dimensional array) data structure, which is highly optimized for efficient mathematical operations. This data structure allows you to store and manipulate large sets of data in a homogeneous and memory-efficient way. 

NumPy is a low-level library written in C for high-level mathematical functions. 
NumPy cleverly overcomes the problem of running slower algorithms on Python using multidimensional arrays and functions that operate on arrays. Any algorithm can then be expressed as a function on these arrays, allowing the algorithms to be run quickly.

Applications of NumPy

- A powerful N-dimensional array object.
- Sophisticated functions.
- Tools for integrating C/C++ and Fortran code.
- Useful linear algebra, Fourier transforms, and random number capabilities.

N-d Array Properties

- a powerful N-dimensional array object
- sophisticated (broadcasting) functions
- tools for integrating C/C++ and Fortran code
- useful linear algebra, Fourier transform, and random number capabilities

Properties of NumPy

- Homogenous (Only one type of data can be stored unlike Python Lists)
- Fixed item size (You can't directly add new elements to an array without creating a new one)

# Creating NumPy Arrays

These are various methods through which we can create NumPy arrays.

1. np.array()
2. np.zeros/ones/empty/random() functions.
4. np.arange()
5. np.linspace()
6. Copy()
7. np.identity()

```python
# creating 1-dimension array (Vector)
arr1 = np.array([1,2,3,4,5],dtype='int16') 
print(arr1)

#dtype changes the datatype at will. Explained later.
```

```python
# creating 2-dimension array (Matrix)
arr2 = np.array([[1,2],[3,4]])
print(arr2)
```

```python
# creating 3-dimension array (Tensor)
arr3 = np.array([[[1,2],[3,4]],[[5,6],[7,8]]],dtype='float64')
print(arr3)
```

```python
# type - Function returns the datatype of the object.
type(arr1) 
```

```python
# dtype - Create array with a custom datatype.
print(np.array([1,2,3,4,5],dtype=float))
print(np.array([1,2,3,4,5],dtype=bool))
print(np.array([1,2,3,4,5],dtype=complex))
```

# Properties and Arrtributes

Some Important Numpy attributes (methods).

1. ndim
2. shape
3. size
4. itemsize
5. dtype
6. astype()

```python
print(arr1)

print('\n')
print(arr2)
print('\n')
print(arr3)
```

```python
# ndim - Dimension of an array.
print(arr1.ndim) #Vector
print(arr2.ndim) #Matrix
print(arr3.ndim) #Tensor
```

```python
# shape - Number of items in rows and columns of the array.
print(arr1.shape) 
print(arr2.shape)
print(arr3.shape)
```

```python
# size - Prints number of elements inside the array. (Simply Multiply the shape.)

print(arr1.size)
print(arr2.size)
```

```python
# itemsize - Prints the size (memory consumption) of one element of that array. 
# Like a float is 8 bytes, sizes may differ in different environment.

print(arr1.itemsize) #int16 - 2 bytes
print(arr3.itemsize) #int64 - 8 bytes
print(arr3.itemsize) #float64 - 8 bytes
```

```python
# dtype - Gives the datatype of the array elements (remember homogenous).

print(arr1.dtype)
print(arr2.dtype)
print(arr3.dtype)
```

# List vs Array

How n-D Array's are better than Lists for large numerical datasets

1. Faster
2. Convenient
3. Less Memory

```python
import sys
list_a = list(range(100))
np_arr = np.arange(100)
```

```python
print(sys.getsizeof(87)*len(list_a)) 
#getsizeof tells the memory consumed by a particular item, 
# here we multiplied that with total number of such items, i.e., 100.
```

```python
print(np_arr.itemsize*np_arr.size) 
# here we can see that we are already saving a 
# lot of space by using numpy arrays.
```
