## What is Pandas?

Pandas is a fast, powerful, flexible and easy to use opensource data analysis and manipulation tool, built on top of the Python programming language.

### Basic data structures in pandas

<b>Pandas provides two types of classes for handling data: </b>

Series: a one-dimensional labeled array holding data of any type
such as integers, strings, Python objects etc.

DataFrame: a two-dimensional data structure that holds data like a two-dimension array or a table with rows and columns. Like in excel.




# Series

## Creating Series

```python
# string
country = pd.Series(['india','germany','usa','japan','russia'])

print(country)

# there is always an auto generated index attached to series.
```

```python
# integers
runs = pd.Series([12,41,25,66,75,34],dtype='uint16')
print(runs)
```

```python
# using custom index
marks = [67,35,86,69]
subjects = ['maths','english','science','hindi']

uc_marks = print(pd.Series(marks,index=subjects,name = "Utkarsh's Marks"))

#We can set custom index names by using index parameter.
#We can assign a referable name to series object by using name parameter.
```

```python
# series from dict

marks = {
    'maths' : 67,
    'english' : 87,
    'science' : 56,
    'hindi' : 90
}
uc_marks = pd.Series(marks,name='Utkarsh ke marks')
print(uc_marks)
```

### drop_duplicates() and duplicated()

Removes the duplicate values.

```python
# drop_duplicates

temp = pd.Series([12,51,73,44,74,12,52,74])
print(temp.drop_duplicates()) # keeps first occurance only
# if you wanna keep last occurance, use drop_duplicates(keep = 'last)
```

```python
bollywood.drop_duplicates() #Only kept 1 movie of every actor
```

```python
# To check if data has duplicates.

temp.duplicated()
```

```python
# To check if data has duplicates.

temp.duplicated().sum() # This means 2 duplicate values
```

```python
# To check if data has duplicates.

temp.duplicated().mean().mean() * 100 # %tage of duplicate values in dataset.
```

### To count number of items we can use: 
- .count() - It will count all the non-missing (non-null) values
- .size - All items (including nulls)

```python
temp = pd.Series([1,2,4,np.nan,8,np.nan,24,2,54,525,25,np.nan,25])
print(temp)
```

```python
temp.size
```

```python
print(temp.count())
```

### isnull()

```python
temp.isnull() # gives a boolean series
```

```python
temp[temp.isnull()] #masking to filer null values 
```

```python
print(temp.isnull().sum()) #number of missing values
print(temp.isnull().mean() * 100) #Percent of missing values from the data
```

### Handling Missing Values

```python
# fillna

temp.fillna(0) # If you don't wanna remove NULL's, and replace with some other value.
```

### apply()

Allows to use custom logic on series

```python
bollywood
```

```python
# using our custom function. 

# Example - keep actor first name only and in all capital letters.
bollywood.apply(lambda x:x.split()[0].upper())
```
