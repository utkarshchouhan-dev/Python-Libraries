### Basic data structures in pandas

<b>Pandas provides two types of classes for handling data: </b>

Series: a one-dimensional labeled array holding data of any type
such as integers, strings, Python objects etc.

DataFrame: a two-dimensional data structure that holds data like a two-dimension array or a table with rows and columns. Like in excel.




# Series

## Creating Series

```python
# string
country = pd.Series(['india','germany','usa','japan','russia'])

print(country)

# there is always an auto generated index attached to series.
```

```python
# integers
runs = pd.Series([12,41,25,66,75,34],dtype='uint16')
print(runs)
```

```python
# using custom index
marks = [67,35,86,69]
subjects = ['maths','english','science','hindi']

uc_marks = print(pd.Series(marks,index=subjects,name = "Utkarsh's Marks"))

#We can set custom index names by using index parameter.
#We can assign a referable name to series object by using name parameter.
```

```python
# series from dict

marks = {
    'maths' : 67,
    'english' : 87,
    'science' : 56,
    'hindi' : 90
}
uc_marks = pd.Series(marks,name='Utkarsh ke marks')
print(uc_marks)
```

## Series Attributes
1. size
2. dtype
3. name
4. index
5. values

```python
country = pd.Series(['india','germany','usa','japan','russia'])

print(country)
```

```python
# shape and size
print(country.shape)
print(country.size)
```

```python
# dtype

print(country.dtype)
```

```python
# name
print(uc_marks.name)
```

```python
# is_unique -> True - If all items are unique, else False

print(uc_marks.is_unique)
print(pd.Series([1,2,3,4,4,1,9]).is_unique)
```

```python
# index

print(uc_marks.index)
print(runs.index)
```

```python
# values -> outputs a numpy array

uc_marks.values
```

## Series Methods

```python
# head
subs.head(3) #default 5
```

```python
# tail
subs.tail() #default 5
```

```python
# tail
kohli.tail(8)
```

```python
# sample -> Randomly selected samples (good if data is biased)
bollywood.sample(5) #Default 1
```

```python
# value_count -> Counts the frequency of every item in the dataset.

bollywood.value_counts() #Number of movies done by each actor (desc order)
```

```python
# sort_values() - Doesn't affects original array. unless inplace = True

print(kohli.sort_values())
print(kohli.sort_values(ascending=False),'\n') #For descending order
print(kohli.sort_values(ascending=False).head(1).values[0])
# kohli.sort_values(inplace=True) # This will change the original array. Use carefully.
```

```python
# sort_index() -> Insead of values it will sort based on index.

print(bollywood.sort_index()) # ascending = False
# bollywood.sort_index(inplace=True) # Changes the original array. Be Careful
```
