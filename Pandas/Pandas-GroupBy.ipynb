# Groupby()

- We use groupby on categorical variables.

```python
import numpy as np
import pandas as pd
```

```python
movies = pd.read_csv('/Users/utkarshchouhan/Desktop/Python-Libraries/Datasets/datasets-groupby/imdb-top-1000.csv')
ipl = pd.read_csv('/Users/utkarshchouhan/Desktop/Python-Libraries/Datasets/datasets-pandas2/ipl-matches.csv')
```

```python
movies.head(3)
```

```python
genres = movies.groupby('Genre')
```

```python
# applying built-in aggregation functions
genres.sum(numeric_only=True).sort_values('No_of_Votes',ascending=False)
```

All min/max/std/sum/prod/var functions will work on this groupby object.

```python
genres.mean(numeric_only=True)
```

```python
genres.std(numeric_only=True)
```

```python
genres.max()
```
