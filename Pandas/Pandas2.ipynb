## What is Dataframe ?

Pandas DataFrame is a way to represent and work with tabular data. It can be seen as a table that organizes data into rows and columns, making it a two-dimensional data structure.

A single row or column in dataframe is called a series.

### Creating DataFrame

```python
# using list

student_data = [
    [100,80,10],
    [90,70,7],
    [120,100,14],
    [80,50,2]
]

student_data = pd.DataFrame(student_data,columns=['iq','marks','package'])
```

```python
# using dict

student_dict = {
    'name' : ['uc','ac','vk','sh'],
    'iq' : [100,90,120,80],
    'marks' : [80,70,100,50],
    'package' : [10,7,14,2]
}

student_dict = pd.DataFrame(student_dict)
student_dict.set_index('name',inplace=True)
```

### .dropna() 

Drops all the rows with a missing value (even a single missing value will remove that row). Works on series and df

```python
student_data.dropna() #default parameter (how = 'any)
```

```python
student_data.dropna(how = 'all') #if all values are missing (how = 'any)
```

```python
# removing rows based on a value from missing column

student_data.dropna(subset=['package']) 
# Will drop only those rows where a missing value occurs in the package column
```

### fillna()

Insead of droping use a value to fill missing data.
Works on both series and df.

```python
movies[movies['poster_path'].isnull()].head(2)
```

```python
# these poster paths are missing, instead of removing these rows lets fill it with something.
movies['poster_path'].fillna('missing_link')
```

missing data can be filled with different strategies.
