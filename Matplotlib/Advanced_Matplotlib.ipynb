## Colored Scatterplots

```python
# Let's plot the sepal length and petal length in a scatter plot.
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'])
plt.xlabel('Sepal Length')
plt.xlabel('Petal Length')

#we know that there were 3 species of iris in the dataset, we can color encode
# them in order to differentiate them.
```

```python
# Let's first create a numerical encoding for the species names

iris['Species_Enc'] = iris['Species'].replace({'Iris-setosa': 0,'Iris-versicolor': 1,'Iris-virginica': 2}).astype('int16')
print(iris.head())
```

```python
# Let's plot the sepal length and petal length in a scatter plot.
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species_Enc'],cmap='winter',alpha=0.5)
plt.xlabel('Sepal Length')
plt.xlabel('Petal Length')
plt.colorbar() # adding a color bar shows, which color represents what value.
# just add the c=dataframe parameter, with cmap='jet', we can use different color maps
# alpha - change the opacity of the graph/plot.
# https://matplotlib.org/stable/users/explain/colors/colormaps.html
```

## Plot size

```python
plt.figure(figsize=(12,5))

plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species_Enc'],cmap='winter',alpha=0.5)
plt.xlabel('Sepal Length')
plt.xlabel('Petal Length')
plt.colorbar()
```

## Annotations

```python
batter = pd.read_csv('/Users/utkarshchouhan/Desktop/Python-Libraries/Datasets/datasets-matplotlib/batter.csv')
sample_df = batter.head(100).sample(25,random_state=5) #random_state uses seed, to create reproducible results.
sample_df
```

```python
# We can use annotations to fix this issue.

plt.figure(figsize=(18,10)) #must be defined before plotting.

plt.scatter(sample_df['avg'],sample_df['strike_rate'])
#here each dot represents a batman, but how do we know which one represents whom.

for i in range(sample_df.shape[0]): #we will loop for each row in the data.
    plt.text(sample_df['avg'].values[i],sample_df['strike_rate'].values[i],sample_df['batter'].values[i])

plt.show()
```

```python
# Changing the size.

plt.figure(figsize=(18,10)) #must be defined before plotting.

plt.scatter(sample_df['avg'],sample_df['strike_rate'],s=sample_df['runs']*0.2)
#here each dot represents a batman, but how do we know which one represents whom.

for i in range(sample_df.shape[0]): #we will loop for each row in the data.
    plt.text(sample_df['avg'].values[i],sample_df['strike_rate'].values[i],sample_df['batter'].values[i])

plt.show()
```

## Horizontal and Vertical Lines

```python
plt.figure(figsize=(18,10)) 
plt.scatter(sample_df['avg'],sample_df['strike_rate'],s=sample_df['runs']*0.2)

for i in range(sample_df.shape[0]): #we will loop for each row in the data.
    plt.text(sample_df['avg'].values[i],sample_df['strike_rate'].values[i],sample_df['batter'].values[i])

# adding axis horizontal and axis vertical lines
plt.axhline(130,color='red')
plt.axhline(140,color='blue',alpha=0.7)
plt.axvline(28,color='green')

plt.show()
```

# Subplots

We can plot multiple plots at once, this is usefull when we are dealing 
with multiple columns and wanna compare the differences.

```python
fig, ax = plt.subplots(figsize=(15,8))
ax.scatter(batter['avg'],batter['strike_rate'],color='darkgreen',marker='+')
ax.set_title("Average vs Strike-rate Plot")
ax.set_xlabel("Average")
ax.set_ylabel("Strike-rate")
fig.show()
```

```python
# Plotting 2 charts side by side.

plt.subplots(nrows=2,ncols=1)
```

```python
fig, ax = plt.subplots(nrows=2,ncols=1,figsize=(15,12),sharex=True)
ax[0].scatter(batter['avg'],batter['strike_rate'],color='darkgreen',marker='+')
ax[1].scatter(batter['avg'],batter['runs'],color='blue',marker='*')


ax[0].set_title("Average vs Strike-rate Plot")
ax[0].set_xlabel("Average")
ax[0].set_ylabel("Strike-rate")

ax[1].set_title("Average vs Runs Plot")
ax[1].set_xlabel("Average")
ax[1].set_ylabel("Runs")

fig.show()
```

```python
# Lets make 4 graphs

fig, ax = plt.subplots(nrows=2,ncols=2,figsize=(18,12))
ax[0,0].scatter(batter['avg'],batter['strike_rate'],color='red',marker='+')
ax[0,1].scatter(batter['avg'],batter['runs'],color='blue',marker='*')
ax[1,0].hist(batter['strike_rate'])
ax[1,1].hist(batter['avg'])

ax[0,0].set_xlabel('Average')
ax[0,0].set_ylabel('Strikerate')

ax[0,1].set_xlabel('Average')
ax[0,1].set_ylabel('Runs')

ax[1,0].set_xlabel('Average')
ax[1,0].set_ylabel('Strike-Rate')

ax[1,1].set_xlabel('Average')
ax[1,1].set_ylabel('Average')
```

```python
#alternate method
fig = plt.figure(figsize=(12,10))
ax1 = fig.add_subplot(2,1,1)
ax1.scatter(batter['avg'],batter['strike_rate'],color='red')

ax2 = fig.add_subplot(2,1,2)
ax2.hist(batter['runs'])
```

```python
fig = plt.figure(figsize=(10,10))

ax = plt.subplot(projection='3d')

ax.scatter3D(batter['runs'],batter['avg'],batter['strike_rate'],color='red',marker='*')
ax.set_title('IPL batsman analysis')
ax.set_xlabel('Runs')
ax.set_ylabel('Avg')
ax.set_zlabel('Strike_rate')
```

### 3d Lineplot

```python
x = [0,1,5]
y = [0,10,15]
z = [0,14,18]

fig = plt.figure(figsize=(10,10))

ax = plt.subplot(projection='3d')

ax.scatter3D(x,y,z,color='red',s=[100,100,100])
ax.plot3D(x,y,z,color='blue')
```
