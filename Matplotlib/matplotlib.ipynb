```python
# Can we change the plot colors.
# Plotting multiple plots at once
plt.plot(batsman['index'],batsman['V Kohli'], color='blue' ,alpha=0.4) #aplha is the intensity/sharpness of the color.
plt.plot(batsman['index'],batsman['RG Sharma'], color='red')

# Above code produces a plain graph, add some details to it.
plt.title('Rohit Sharma vs Virat Kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel("Runs Scored")
```

```python
# We can change solid lines with linestyle and linewidth to change the thickness

# Can we change the plot colors.

# Plotting multiple plots at once
plt.plot(batsman['index'],batsman['V Kohli'], color='green',alpha=0.6,linestyle='dotted',linewidth=3)
plt.plot(batsman['index'],batsman['RG Sharma'], color='red',linestyle='dashed') #dashdot

# Above code produces a plain graph, add some details to it.
plt.title('Rohit Sharma vs Virat Kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel("Runs Scored")
```

```python
# Show the markers in the 2d plot. https://matplotlib.org/stable/api/markers_api.html

plt.plot(batsman['index'],batsman['V Kohli'], color='blue',linestyle='dotted',linewidth=3, marker='D',markersize=4)
plt.plot(batsman['index'],batsman['RG Sharma'], color='red',linestyle='dashed', marker='*',markersize=8)

plt.title('Rohit Sharma vs Virat Kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel("Runs Scored")
```

```python
# add a legend to each line, a legend. 

plt.plot(batsman['index'],batsman['V Kohli'], color='blue',linestyle='dotted',linewidth=3, marker='X',markersize=4,label='Kohli')
plt.plot(batsman['index'],batsman['RG Sharma'], color='red',linestyle='dashed', marker='x',markersize=8,label='Sharma')

plt.title('Rohit Sharma vs Virat Kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel("Runs Scored")


# add this to display thevlegend
plt.legend() 
#default position is best, but can be changed with plt.legend(loc='center left') 
```

```python
# limiting axes problem

price = [51000,48000,51000,59000,54000,5200000] #because of an outlier the graph became a flatline. and those minor variations are hard to capture.
year = [2015,2016,2017,2018,2019,2020]

plt.plot(year,price)
```

```python
# limiting axes solution

price = [51000,48000,51000,59000,54000,5200000] 
year = [2015,2016,2017,2018,2019,2020]

plt.plot(year,price)
plt.ylim(40000,75000) #now it trim the y axes so that you can see the small variations
plt.xlim(2015, 2018)
```

```python
# To convey more details, use plt.grid()

plt.plot(batsman['index'],batsman['V Kohli'], color='blue',linestyle='dotted',linewidth=3, marker='X',markersize=4,label='Kohli')
plt.plot(batsman['index'],batsman['RG Sharma'], color='red',linestyle='dashed', marker='x',markersize=8,label='Sharma')

plt.title('Rohit Sharma vs Virat Kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel("Runs Scored")
plt.legend() 


# add this to display the grid
plt.grid()
```

### Barchart

We keep a categorical variable on the x-axis and usually keep an aggregated value on the y-axis.

- Univariate and Bivariate analysis
- Numerical(usually an aggregate function) vs categorical
- Use case - Aggregate analysis of groups.

```python
# Count of children that like a particular color
children_count = [10,20,15,40,25,18]
color = ['red', 'blue', 'pink','green','black','white']

plt.bar(color,children_count)
plt.title('Count of children that like a particular color')
```

```python
# We can also plot a horizontal barchart with barh function.

plt.barh(color,children_count)

# If number of categories are less then use horizontal, else go with vertical.
```

```python
bat_score = pd.read_csv('/Users/utkarshchouhan/Desktop/Python-Libraries/Datasets/datasets-matplotlib/batsman_season_record.csv')
bat_score.head(10)
```

```python
plt.bar(np.arange(bat_score.shape[0]) - 0.2,bat_score['2015'], width=0.2)
plt.bar(np.arange(bat_score.shape[0]),bat_score['2016'], width=0.2)
plt.bar(np.arange(bat_score.shape[0]) + 0.2,bat_score['2017'], width=0.2)

plt.xticks(np.arange(bat_score.shape[0]),bat_score['batsman'])


plt.xlabel('Batsman Name')
plt.ylabel('Runs scored')
```

```python
# Stacked Bar Chart

plt.bar(bat_score['batsman'], bat_score['2015'], label= '2015')
plt.bar(bat_score['batsman'], bat_score['2016'], bottom= bat_score['2015'], label= '2016')
plt.bar(bat_score['batsman'], bat_score['2017'], bottom= bat_score['2015'] + bat_score['2016'], label= '2017')

plt.legend()
plt.show()
```

```python
fig = plt.figure()
ax1 = fig.add_subplot(2, 2, 1)
ax2 = fig.add_subplot(2, 2, 2)
ax3 = fig.add_subplot(2, 2, 3)
ax1.hist(np.random.standard_normal(100), bins=20, color="black", alpha=0.4)
ax2.scatter(np.arange(30), np.arange(30) + 3 * np.random.standard_normal(30))
ax3.plot(np.random.standard_normal(50).cumsum(), color="black",linestyle="dashed")
```
