```python
# from pandas dataframe
plt.plot(batsman['index'],batsman['V Kohli'])
```

```python
# Plotting multiple plots at once
plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])

# Above code produces a plain graph, add some details to it.
plt.title('Rohit Sharma vs Virat Kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel("Runs Scored")
```

```python
# Can we change the plot colors.

# Plotting multiple plots at once
plt.plot(batsman['index'],batsman['V Kohli'], color='blue')
plt.plot(batsman['index'],batsman['RG Sharma'], color='red')

# Above code produces a plain graph, add some details to it.
plt.title('Rohit Sharma vs Virat Kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel("Runs Scored")
```

```python
# We can change solid lines with linestyle and linewidth to change the thickness

# Can we change the plot colors.

# Plotting multiple plots at once
plt.plot(batsman['index'],batsman['V Kohli'], color='blue',linestyle='dotted',linewidth=3)
plt.plot(batsman['index'],batsman['RG Sharma'], color='red',linestyle='dashed')

# Above code produces a plain graph, add some details to it.
plt.title('Rohit Sharma vs Virat Kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel("Runs Scored")
```

```python
# Show the markers in the 2d plot.

plt.plot(batsman['index'],batsman['V Kohli'], color='blue',linestyle='dotted',linewidth=3, marker='D',markersize=4)
plt.plot(batsman['index'],batsman['RG Sharma'], color='red',linestyle='dashed', marker='D',markersize=8)

plt.title('Rohit Sharma vs Virat Kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel("Runs Scored")
```

```python
# add a legend to each line

plt.plot(batsman['index'],batsman['V Kohli'], color='blue',linestyle='dotted',linewidth=3, marker='D',markersize=4,label='Kohli')
plt.plot(batsman['index'],batsman['RG Sharma'], color='red',linestyle='dashed', marker='D',markersize=8,label='Sharma')

plt.title('Rohit Sharma vs Virat Kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel("Runs Scored")
plt.legend()
```
